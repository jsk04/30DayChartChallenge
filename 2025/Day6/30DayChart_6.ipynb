```python
import pandas as pd

injuries_df = pd.read_csv("football_injury.csv")
injuries_df.head()
```

```python
injuries_df['Year'] = injuries_df['Year'].astype(str)
```

```python
import plotly.graph_objects as go
```

```python
fig = go.Figure()

fig.add_trace(go.Barpolar(
    r = list(injuries_df['Game']),
    theta = list(injuries_df['Year']),
    name = 'Injuries during Game',
    marker_color = 'rgb(46,109,255)',
    marker_line_color = 'black',
    hovertemplate="<b>Injuries: %{r}<b>",
    opacity = 0.7

))
fig.add_trace(go.Barpolar(
    r = list(injuries_df['Practice']),
    theta = list(injuries_df['Year']),
    name = 'Injuries during Practice',
    marker_color = 'rgb(139,222,52)',
    marker_line_color = 'black',
    hovertemplate="<b>Injuries: %{r}<b>",
    opacity = 0.7

))
fig.update_layout(
    title='Football Injuries during Game/Practice, 2015-2024',
    font_size = 12,
    legend_font_size=15,
    polar_angularaxis_rotation=90,
    width=900,
    height=900,
    polar = dict(
        bgcolor = "rgb(223, 223, 223)",
        angularaxis = dict(
            linewidth = 3,
            showline = True,
            linecolor = "black"
        )
        # radialaxis = dict(
        #     linewidth = 2,
        #     showline = True,
        #     gridcolor = "black",
        #     gridwidth=2
        # )
    )
)
fig.show()
```

```python
fig.write_image("30DayChart-6.png")
fig.write_html("30DayChart-6.html")
```
