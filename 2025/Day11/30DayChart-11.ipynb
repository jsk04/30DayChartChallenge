```python
import pandas as pd
import matplotlib.pyplot as plt 
import gzip
import csv
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
```

```python
rain = pd.read_csv('weather.csv', usecols=['DATE', 'PRCP'])
```

```python
rain = rain.dropna(subset='PRCP')
```

```python
rain['y'] = 1
```

```python
fig = px.bar(
    rain,
    x="DATE",
    y="y",
    color="PRCP",
    color_continuous_scale="YlGnBu",
    hover_data={"PRCP": True, "DATE": True, "y": False},
    labels={"PRCP": "Precipitation (in)", "y": ""},
    height=600  # Control the overall stripe height
)

# Remove unnecessary stuff for a clean stripe look
fig.update_layout(
    xaxis=dict(
        showticklabels=False,
        showgrid=False,
        ticks=''
    ),
    showlegend=False,
    yaxis=dict(showticklabels=False, showgrid=False, zeroline=False),
    # xaxis=dict(title="Date"),
    bargap=0,  # No gap between bars
    margin=dict(t=100, b=100, l=20, r=20),
    coloraxis_showscale=False,
    plot_bgcolor='white',
    paper_bgcolor='white',
    title=dict(text="Daily Precipitation in Frankfort, KY from Jan-Apr 2025", x=0.0105),
    annotations=[
        dict(
            text="Recorded from Station USC00153025",
            xref="paper", yref="paper",
            x=-0.02, y=1.10,
            showarrow=False,
            font=dict(size=12),
            font_color='grey'
        ),
        dict(
            text="Source: NOAA",
            xref="paper", yref="paper",
            x=0.5, y=-0.16,
            showarrow=False,
            font=dict(size=10),
            font_color='grey'
        ),
        dict(
            text="January",
            xref="paper", yref="paper",
            x=0.005, y=-0.06,
            showarrow=False,
            font=dict(size=12),
            font_color='black'
        ),
        dict(
            text="April",
            xref="paper", yref="paper",
            x=1, y=-0.06,
            showarrow=False,
            font=dict(size=12),
            font_color='black'
        ),
    ]
)

# Make the bars look flat like stripes
fig.update_traces(
    marker_line_width=0
)

fig.show()
```

```python
fig.write_image("30DayChart-11.png")
fig.write_html("30DayChart-11.html")
```
